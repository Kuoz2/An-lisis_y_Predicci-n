```python
import pandas as pd
import matplotlib.pyplot as plt
import io
```

```python
data = """
fecha,ventas
2023-01-01,200
2023-01-02,220
2023-01-03,210
2023-01-04,205
2023-01-05,215
2023-01-06,225
"""

# Cargar los datos en un DataFrame de Pandas
df = pd.read_csv(io.StringIO(data), sep=",")
```

```python
df['fecha'] = pd.to_datetime(df['fecha'])

plt.style.use('seaborn-v0_8')
```

```python
# Gráfico de líneas de las ventas
plt.figure(figsize=(10, 6))
plt.plot(df['fecha'], df['ventas'])
plt.title("Ventas Diarias")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Crear un gráfico de dispersión
plt.figure(figsize=(10, 4))
plt.scatter(df['fecha'], df['ventas'], color='red', s=100, edgecolors='black', label='Puntos de venta')
plt.title("Ventas vs Fecha")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.show()
```

```python
# Crear un histograma de las ventas
plt.figure(figsize=(10, 4))
plt.hist(df['ventas'], bins=5, color='skyblue', edgecolor='black', alpha=0.8)
plt.title("Distribución de Ventas")
plt.xlabel("Rango de ventas")
plt.ylabel("Frecuencia en dìas")
plt.show()
```
