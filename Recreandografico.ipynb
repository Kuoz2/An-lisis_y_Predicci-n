```python
import seaborn as sns
import matplotlib.pyplot as plt


```

```python
# Cargar el dataset
df = sns.load_dataset('tips')
```

```python
df
```

```python
fig, ax = plt.subplots(figsize=(24, 6))

# Gráfico de dispersión con cuadrícula, leyenda, anotaciones y flechas
# Rellenamos con: time, smoker, 100, ax
# Nota: Para los colores de la imagen usamos palette=['#E74C3C', '#3498DB']
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', style='smoker', s=100, palette=['#E74C3C', '#3498DB'], ax=ax)

ax.set_title('Relación entre Total Bill y Tip con Anotaciones y Flechas', fontsize=20)
ax.set_xlabel('Total Bill ($)', fontsize=15)
ax.set_ylabel('Tip ($)', fontsize=15)
ax.legend(loc='upper right', fontsize='large', shadow=False, title='Detalles', frameon=False)

# Añadir cuadrícula
# Rellenamos con: 'both', '--', 0.5, 'gray'
ax.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')

# Añadir anotación
# El punto más alto de Dinner es (50.81, 10.0)
# Rellenamos xy=(50.81, 10), xytext=(35, 10)
ax.annotate('Mayor tip en la cena', xy=(50.81, 10), xytext=(50, 15.5),
            arrowprops=dict(facecolor='black', shrink=0.05))

# Añadir flecha
# La flecha en la imagen sale de un punto medio hacia arriba
# Rellenamos con: 35, 5.5, 12, 3 (inicio x, inicio y, desplazamiento dx, desplazamiento dy)
ax.arrow(37, 6, 5, 2, head_width=0.8, head_length=0.5, fc='green', ec='green')

# Mostrar el gráfico
plt.show()
```
