```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
dk_mineria= pd.read_csv("/content/drive/MyDrive/CURSOPYTHON/MiningProcess_Flotation_Plant_Database.csv")

dk_mineria.head()
```

```python

dk_mineria['date']= pd.to_datetime(dk_mineria['date'], errors='coerce')

columnas_sensores= dk_mineria.columns.drop('date')
for col in columnas_sensores:
    dk_mineria[col]= dk_mineria[col].astype(str).str.replace(',', '.')
    dk_mineria[col]= pd.to_numeric(dk_mineria[col], errors='coerce')
    print(f"Transformación completada. Tamaño final: {dk_mineria.shape[0]} filas y {dk_mineria.shape[1]} columnas.")
```

```python
nulos= dk_mineria.isnull().sum()
if nulos.sum() > 0:
    print("Se detectaron los siguientes nulos tras la conversión:")
    display(nulos[nulos > 0])
else:
    print("El dataset no contiene valores nulos tras la limpieza.")

display(dk_mineria.describe().round(2))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.histplot(dk_mineria['% Silica Concentrate'], kde=True, bins=50, ax=axes[0], color='teal')
axes[0].set_title('Distribución del % de Sílice en el Concentrado')
axes[0].set_xlabel('% Sílice')
axes[0].set_ylabel('Frecuencia')
sns.boxplot(x=dk_mineria['% Silica Concentrate'], ax=axes[1], color='lightseagreen')
axes[1].set_title('Detección de Outliers en la Variable Objetivo')
axes[1].set_xlabel('% Sílice')

plt.tight_layout()
plt.show()
```

```python
matriz_corr= dk_mineria.drop(columns=['date']).corr()
corr_objetivo= matriz_corr['% Silica Concentrate'].sort_values(ascending=False)
display(corr_objetivo.head(8))
```
