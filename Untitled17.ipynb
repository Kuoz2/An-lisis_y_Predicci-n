```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dk= pd.read_csv("/content/drive/MyDrive/CURSOPYTHON/insurance.csv")
```

```python
mediana_cargo= dk['charges'].median()
dk['costo_alto']= (dk['charges'] > mediana_cargo).astype(int)
```

```python
dk= dk.drop('charges', axis=1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline
```

```python
x= dk.drop('costo_alto', axis=1)
y= dk['costo_alto']
num_cols = x.select_dtypes(include=['int64', 'float64']).columns
cat_cols= x.select_dtypes(include=['object']).columns

num_tranfor= Pipeline(steps= [
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

cat_transformer= Pipeline(steps= [
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(drop= 'first', sparse_output= False))
])

preposor=  ColumnTransformer(transformers= [
    ('num', num_tranfor, num_cols ),
    ('cat', cat_transformer, cat_cols)
])

x_processed= preposor.fit_transform(x)

x_train, x_test, y_train, y_test= train_test_split(x_processed, y, test_size= 0.2, random_state= 42, stratify= y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
```

```python
modelos= {
    'Regresion Logistica': LogisticRegression(random_state=42),
    'KNN': KNeighborsClassifier(),
    'Decision Tree': DecisionTreeClassifier(random_state=42)
}

for nombre, modelo in modelos.items():
  score= cross_val_score(modelo, x_train, y_train, cv=5, scoring= 'accuracy')
  print(f"{nombre}: Accuracy Promedio = {score.mean():.4f} (Desviación = {score.std():.4f})")
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
params_grid={
    'max_depth': [None, 5, 10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'criterion': ['gini', 'entropy']
}

dt_b= DecisionTreeClassifier(random_state=42)
grid_search= GridSearchCV(estimator= dt_b, param_grid= params_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(x_train, y_train)

print("--- Resultados GridSearchCV ---")
print("Mejores parámetros:", grid_search.best_params_)
print("Mejor accuracy:", grid_search.best_score_)
```

```python
random_search= RandomizedSearchCV(estimator=dt_b, param_distributions=param_grid,
                                   n_iter=15, cv=5, scoring='accuracy',
                                   random_state=42, n_jobs=-1)
random_search.fit(x_train, y_train)

print("--- Resultados RandomizedSearchCV ---")
print("Mejores parámetros:", random_search.best_params_)
print("Mejor accuracy:", random_search.best_score_)

mejor_modelo= grid_search.best_estimator_
```

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix, roc_curve)
```

```python
y_pred= mejor_modelo.predict(x_test)
y_pred_proba= mejor_modelo.predict_proba(x_test)[:,1]
```

```python
# 1. Métricas de Rendimiento
print("--- Métricas de Rendimiento del Modelo Optimizado ---")
print(f"Exactitud (Accuracy): {accuracy_score(y_test, y_pred):.4f}")
print(f"Precisión (Precision): {precision_score(y_test, y_pred):.4f}")
print(f"Sensibilidad (Recall): {recall_score(y_test, y_pred):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred):.4f}")
print(f"ROC-AUC: {roc_auc_score(y_test, y_pred_proba):.4f}")
```

```python
cm= confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Costo Bajo', 'Costo Alto'], yticklabels=['Costo Bajo', 'Costo Alto'])
plt.title("Matriz de Confusión")
plt.xlabel("Predicciones")
plt.ylabel("Etiqueta real")
plt.show()
```

```python
fpr, tpr, thresholds= roc_curve(y_test, y_pred_proba)
plt.figure(figsize=(6,4))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'Curva ROC (AUC = {roc_auc_score(y_test, y_pred_proba):.4f})')
plt.plot([0,1], [0,1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Tasa de Falsos Positivos')
plt.ylabel('Tasa de Verdaderos Positivos')
plt.title('Característica Operativa del Receptor (ROC)')
plt.legend(loc='lower right')
plt.show()
```
